# **5) Graphical Analysis**

### 1. Bar Plot: Distribution of Crops

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=Data, x='Crop', order=Data['Crop'].value_counts().index, palette='viridis')
plt.title('Distribution of Crops')
plt.xlabel('Crop Type')
plt.ylabel('Number of Farms')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### 2. Histogram: Distribution of Rainfall

```python
plt.figure(figsize=(10, 6))
sns.histplot(Data['Rainfall_mm'].dropna(), bins=30, kde=True, color='skyblue')
plt.title('Distribution of Rainfall (mm)')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

### 3. Line Plot: Average Yield Over Seasons

```python
seasonal_yield = Data.groupby('Season')['Yield_Tonnes_Ha'].mean().reindex(['Kharif', 'Rabi', 'Zaid'])

plt.figure(figsize=(10, 6))
sns.lineplot(x=seasonal_yield.index, y=seasonal_yield.values, marker='o', color='green')
plt.title('Average Yield (Tonnes/Ha) Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 4. Scatter Plot: Temperature vs. Yield

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=Data, x='Avg_Temperature_C', y='Yield_Tonnes_Ha', hue='Season', palette='deep', alpha=0.7)
plt.title('Average Temperature (C) vs. Yield (Tonnes/Ha)')
plt.xlabel('Average Temperature (C)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.tight_layout()
plt.show()
```

### 5.Box Plot: Yield Distribution by Season

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=Data, x='Season', y='Yield_Tonnes_Ha', palette='pastel')
plt.title('Yield Distribution by Season')
plt.xlabel('Season')
plt.ylabel('Yield (Tonnes/Ha)')
plt.tight_layout()
plt.show()
```

### 6. Correlation Heatmap of Numeric Columns

```python
# Select only numeric columns for correlation analysis
numeric_cols = Data.select_dtypes(include=np.number)

# Calculate the correlation matrix
correlation_matrix = numeric_cols.corr()

# Plotting the heatmap
plt.figure(figsize=(16, 12))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Heatmap of Numeric Columns')
plt.tight_layout()
plt.show()
```

### 7. Seasonal Variation in Resource Usage

```python
# Calculate the average profit per crop per season
seasonal_crop_profit = Data.groupby(['Season', 'Crop'])['Profit_INR'].mean().unstack()

plt.figure(figsize=(14, 8))
sns.heatmap(seasonal_crop_profit, annot=True, fmt=".0f", cmap="YlGnBu", linewidths=.5)
plt.title('Average Profit (INR) per Crop Type Across Seasons')
plt.xlabel('Crop Type')
plt.ylabel('Season')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### 8. Average Profit Trends for Top 3 Crops by Season

```python
# Get the top 3 most profitable crops
top_3_crops = top_5_profitable_crops.head(3).index.tolist()

# Filter seasonal_crop_profit for these top 3 crops
seasonal_profit_top_3 = seasonal_crop_profit[top_3_crops]

# Prepare data for plotting (unstacking makes 'Crop' a column)
seasonal_profit_plot_data = seasonal_profit_top_3.reset_index().melt(id_vars='Season', var_name='Crop', value_name='Average_Profit_INR')

plt.figure(figsize=(12, 7))
sns.barplot(data=seasonal_profit_plot_data, x='Season', y='Average_Profit_INR', hue='Crop', palette='viridis')
plt.title('Average Profit (INR) for Top 3 Crops Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
plt.tight_layout()
plt.show()
```

### 9. Pesticide Usage Patterns for Selected Crops by Season

```python
crops_to_analyze = ['Sugarcane', 'Chilli', 'Cotton']

# Filter data for the specified crops
pesticide_usage_filtered = Data[Data['Crop'].isin(crops_to_analyze)]

# Calculate average pesticide usage per crop per season
seasonal_pesticide_usage = pesticide_usage_filtered.groupby(['Season', 'Crop'])['Pesticide_Litre_ha'].mean().reset_index()

plt.figure(figsize=(12, 7))
sns.barplot(data=seasonal_pesticide_usage, x='Season', y='Pesticide_Litre_ha', hue='Crop', palette='rocket')
plt.title('Average Pesticide Usage (Litre/ha) for Sugarcane, Chilli, and Cotton Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Pesticide Usage (Litre/ha)')
plt.tight_layout()
plt.show()
```

### 10. Correlation between Pesticide Usage and Profit per Crop

```python
correlation_pesticide_profit = Data.groupby('Crop')[['Pesticide_Litre_ha', 'Profit_INR']].corr().unstack().iloc[:, 1]
correlation_pesticide_profit = correlation_pesticide_profit.sort_values(ascending=False)

print("Correlation between Pesticide Usage and Profit for each Crop:\n")
display(correlation_pesticide_profit)

plt.figure(figsize=(10, 6))
sns.barplot(x=correlation_pesticide_profit.index, y=correlation_pesticide_profit.values, palette='coolwarm')
plt.title('Correlation between Pesticide Usage and Profit per Crop')
plt.xlabel('Crop Type')
plt.ylabel('Correlation Coefficient')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
